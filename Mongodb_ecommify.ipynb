````python
# Insertar los 1000 registros

db = client["ecommify_db"]
product_catalog = db["product_catalog"]

product_catalog.delete_many({})

categories = ["Tecnología", "Ropa", "Hogar", "Deportes", "Libros"]
regions = ["CO-BOG", "CO-ANT", "CO-ATL", "CO-VAC", "CO-SAN"]

def generate_attributes(category):
    if category == "Tecnología":
        return {
            "ram": choice(["8GB", "16GB", "32GB"]),
            "cpu": choice(["Intel i5", "Intel i7", "Intel i9", "AMD Ryzen 7"]),
            "storage": choice(["256GB SSD", "512GB SSD", "1TB SSD"]),
            "gpu": choice(["Integrada", "RTX 3060", "RTX 4080"])
        }
    elif category == "Ropa":
        return {
            "size": choice(["S", "M", "L", "XL"]),
            "color": choice(["Negro", "Blanco", "Azul", "Rojo"]),
            "material": choice(["Algodón", "Poliéster", "Denim"])
        }
    elif category == "Hogar":
        return {
            "material": choice(["Madera", "Metal", "Plástico"]),
            "dimensions": choice(["Pequeño", "Mediano", "Grande"]),
            "brand": choice(["Homecenter", "Ikea", "Haceb"])
        }
    elif category == "Deportes":
        return {
            "sport_type": choice(["Running", "Fútbol", "Ciclismo", "Gimnasio"]),
            "size": choice(["S", "M", "L"]),
            "brand": choice(["Nike", "Adidas", "Puma"])
        }
    else:
        return {
            "author": choice(["Gabriel García Márquez", "Isabel Allende", "Mario Vargas Llosa"]),
            "language": choice(["Español", "Inglés"]),
            "format": choice(["Físico", "Digital"])
        }


````

````python
documents = []

for i in range(1, 1001):
    category = choice(categories)

    product = {
        "_id": f"PROD-{i:04d}",
        "name": f"{category} Producto {i}",
        "description": f"Descripción del producto {i} de la categoría {category}",
        "category": category,
        "price": randint(20, 5000),
        "currency": "USD",
        "seller_id": f"SELLER-{randint(1, 50):03d}",
        "seller_region": choice(regions),
        "attributes": generate_attributes(category),
        "tags": [category.lower(), "ecommify", "catalogo"],
        "images": [f"img_{i}_1.jpg", f"img_{i}_2.jpg"],
        "ratings": {
            "average": round(uniform(1, 5), 1),
            "count": randint(0, 1000)
        },
        "metrics": {
            "total_units_sold": randint(0, 5000),
            "views_count": randint(0, 100000),
            "conversion_rate": round(uniform(0.01, 0.20), 2)
        },
        "recent_reviews": [
            {
                "user_id": randint(1, 5000),
                "rating": randint(1, 5),
                "comment": "Comentario de ejemplo sobre el producto."
            }
        ],
        "created_at": datetime.now(),
        "updated_at": datetime.now()
    }

    documents.append(product)

product_catalog.insert_many(documents)

print("Total productos insertados:", product_catalog.count_documents({}))
````

````python
# Promedio de precio por categoría

pipeline = [
    {
        "$group": {
            "_id": "$category",
            "average_price": {"$avg": "$price"},
            "total_products": {"$sum": 1}
        }
    },
    {
        "$sort": {"average_price": -1}
    }
]

for result in product_catalog.aggregate(pipeline):
    print(result)
````

````python
# Concentración por categoría

category_distribution = list(product_catalog.aggregate([
    {
        "$group": {
            "_id": "$category",
            "total_products": {"$sum": 1}
        }
    }
]))

total = sum(item["total_products"] for item in category_distribution)

hhi_category = sum((item["total_products"] / total) ** 2 for item in category_distribution)

print("Distribución por categoría:")
for item in category_distribution:
    participation = item["total_products"] / total
    print(item["_id"], item["total_products"], round(participation * 100, 2), "%")

print("HHI categoría:", round(hhi_category, 4))
````

````python
# Concentración por región

region_distribution = list(product_catalog.aggregate([
    {
        "$group": {
            "_id": "$seller_region",
            "total_products": {"$sum": 1}
        }
    }
]))

total = sum(item["total_products"] for item in region_distribution)

hhi_region = sum((item["total_products"] / total) ** 2 for item in region_distribution)

print("Distribución por región:")
for item in region_distribution:
    participation = item["total_products"] / total
    print(item["_id"], item["total_products"], round(participation * 100, 2), "%")

print("HHI región:", round(hhi_region, 4))
````

## Sharding teórico para `user_behavior`

La colección `user_behavior` almacena eventos de navegación y comportamiento de usuarios. Esta colección puede crecer de forma masiva debido a eventos como visualización de productos, búsquedas, adición al carrito y navegación por categorías.

En un ambiente productivo con sharding habilitado, se crearía el siguiente índice:

```javascript
db.user_behavior.createIndex({
  "userId": "hashed",
  "period": 1
})

````python
from random import randint, choice

user_behavior = db["user_behavior"]

user_behavior.delete_many({})

events = []

event_types = ["VIEW_PRODUCT", "ADD_TO_CART", "SEARCH", "VIEW_CATEGORY"]

for i in range(1, 501):
    document = {
        "_id": f"USER-{i:04d}-2026-06",
        "userId": randint(1, 200),
        "period": "2026-06",
        "events": [
            {
                "type": choice(event_types),
                "productId": f"PROD-{randint(1, 1000):04d}",
                "timestamp": datetime.now()
            }
        ]
    }
    events.append(document)

user_behavior.insert_many(events)

print("Total documentos user_behavior:", user_behavior.count_documents({}))
````

````python
# Índice equivalente para user_behavior usando PyMongo

user_behavior.create_index([
    ("userId", "hashed"),
    ("period", 1)
])

print("Índice teórico de shard key para user_behavior creado correctamente")
````
